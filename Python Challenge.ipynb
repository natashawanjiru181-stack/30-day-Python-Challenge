```python
# Types of operators
# 1. Arithmetic Operators
# Arithmetic operations: +, -, *, /, %, **, //
x = 10
y = 3

print(x + y)   # 13
print(x - y)   # 7
print(x * y)   # 30
print(x / y)   # 3.333...
print(x % y)   # 1 (remainder)
print(x ** y)  # 1000 (exponent)
print(x // y)  # 3 (floor division)
```

```python
# Backslash (\\) in Python - Summary

# 1. Used as an escape character inside strings.
#    Example: "\\" prints a single backslash.

# 2. Required in Windows file paths to escape each backslash.
#    Example: "C:\\Users\\Natasha\\Documents"

# 3. Used for line continuation when placed at the end of a line.
#    Example:
#    total = 10 + 20 + \
#            30 + 40

# 4. Escapes quotes inside strings so they don't terminate the string.
#    Example: "He said \"Hello\""

# 5. If followed by a normal character (not a special escape), Python prints both.
#    Example: "Hello\\p" → prints Hello\p
```

```python
#Single quote
#usually one part of the pair only
#used when writing short forms
print('You don\'t have responsibilities, there is no reason for you to fail')
```

```python
# Single Quotes vs Double Quotes in Python
# In Python, single quotes (' ') and double quotes (" ") work the SAME.
# Both create string values.
single_quoted = 'Hello'
double_quoted = "Hello"

# The only real difference is:
# You must escape the SAME type of quote used to wrap the string.

# 1. Escaping single quotes inside single-quoted strings
example_single = 'It\'s a good day'  # Must escape the single quote

# 2. Escaping double quotes inside double-quoted strings
example_double = "He said \"Hello\""  # Must escape the double quotes

# 3. Using the opposite quote to avoid escaping
# Using DOUBLE quotes avoids escaping a single quote:
clean_single = "It's a good day"

# Using SINGLE quotes avoids escaping double quotes:
clean_double = 'He said "Hello"'

print(example_single)
print(example_double)
print(clean_single)
print(clean_double)
```

```python
#Join the following lists
front_end = ['HTML', 'CSS', 'JS', 'React', 'Redux']
back_end = ['Node','Express', 'MongoDB']
print(f'{front_end}: {' '.join(back_end)}')#added backend to the frontend list
```

```python
#Join the list in a variable
full_stack=front_end+back_end
print(full_stack)
```
